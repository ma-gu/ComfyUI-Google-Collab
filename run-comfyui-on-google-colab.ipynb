```python
# Clone ComfyUI custom nodes
# Install ComfyUI custom nodes
import pathlib, subprocess

root = pathlib.Path("/content/ComfyUI")
custom = root / "custom_nodes"
custom.mkdir(parents=True, exist_ok=True)

%cd /content/ComfyUI/custom_nodes

repos = [
    "https://github.com/Alectriciti/comfyui-adaptiveprompts.git",
    "https://github.com/scottmudge/ComfyUI-NAG.git",
    "https://github.com/melMass/comfy_mtb.git",
    "https://github.com/rgthree/rgthree-comfy.git",
    "https://github.com/kijai/ComfyUI-WanVideoWrapper.git",
    "https://github.com/Kosinkadink/ComfyUI-VideoHelperSuite.git",
    "https://github.com/Fannovel16/ComfyUI-Frame-Interpolation.git",
    "https://github.com/kijai/ComfyUI-KJNodes.git",
    "https://github.com/yolain/ComfyUI-Easy-Use.git",
    "https://github.com/Suzie1/ComfyUI_Comfyroll_CustomNodes.git",
    "https://github.com/city96/ComfyUI-GGUF.git",
    "https://github.com/Smirnov75/ComfyUI-mxToolkit.git",
    "https://github.com/Comfy-Org/ComfyUI-Manager.git",
    "https://github.com/Artificial-Sweetener/comfyui-WhiteRabbit.git",
    "https://github.com/pythongosssss/ComfyUI-Custom-Scripts.git",
    "https://github.com/filliptm/ComfyUI_Fill-Nodes.git",
    "https://github.com/ashtar1984/comfyui-find-perfect-resolution.git",
    "https://github.com/darksidewalker/ComfyUI-DaSiWa-Nodes.git"
]

# Клонирование
for url in repos:
    name = url.rsplit("/", 1)[-1].replace(".git", "")
    if (custom / name).exists():
        print(f"Skip, already exists: {name}")
        continue
    subprocess.run(["git", "clone", url], check=True)

# Установка requirements (если есть)
for url in repos:
    name = url.rsplit("/", 1)[-1].replace(".git", "")
    req = custom / name / "requirements.txt"
    if req.exists():
        print(f"Installing requirements for {name}")
        subprocess.run(["pip", "install", "-r", str(req)], check=True)
    else:
        print(f"No requirements.txt for {name}, skip")
```

```python
# Download Wan / GGUF / LoRA models for ComfyUI (Civitai через ?token=)

import json
import pathlib
import subprocess
from urllib.parse import urlparse, parse_qs, urlencode, urlunparse
from google.colab import userdata
import ipywidgets as widgets
from IPython.display import display, clear_output

CIVITAI_TOKEN = userdata.get("CIVITAI_TOKEN").strip()
if not CIVITAI_TOKEN:
    print("WARNING: CIVITAI_TOKEN не найден.")

root = pathlib.Path("/content/ComfyUI")
state_dir = root / "output"
state_dir.mkdir(parents=True, exist_ok=True)
state_file = state_dir / "download_ui_state.json"

paths = {
    "vae": root / "models" / "vae",
    "text_encoders": root / "models" / "text_encoders",
    "diffusion": root / "models" / "diffusion_models",
    "loras": root / "models" / "loras",
    "clip_vision": root / "models" / "clip_vision",
    "upscale_models": root / "models" / "upscale_models",
}
for p in paths.values():
    p.mkdir(parents=True, exist_ok=True)


def add_token_to_civitai_url(url: str, token: str) -> str:
    if not token:
        return url
    if "civitai.com" not in url and "civitai.red" not in url:
        return url
    parsed = urlparse(url)
    qs = parse_qs(parsed.query)
    if "token" in qs:
        return url
    qs["token"] = [token]
    return urlunparse(parsed._replace(query=urlencode(qs, doseq=True)))


def build_aria2_cmd(url: str, out_path: pathlib.Path):
    url_with_token = add_token_to_civitai_url(url, CIVITAI_TOKEN)
    return ["aria2c", "-x", "16", "-s", "16", "-o", out_path.name, "-d", str(out_path.parent), url_with_token]


def load_state():
    if not state_file.exists():
        return {}
    try:
        return json.loads(state_file.read_text(encoding="utf-8"))
    except Exception:
        return {}


def save_state(common_checks, checkpoint_radio, lora_checks):
    data = {
        "common": {k: cb.value for k, cb in common_checks.items()},
        "checkpoint": checkpoint_radio.value,
        "loras": {k: cb.value for k, cb in lora_checks.items()},
    }
    state_file.write_text(json.dumps(data, ensure_ascii=False, indent=2), encoding="utf-8")


def download_one(url, out_path):
    if out_path.exists():
        print(f"Already exists, skip: {out_path.name}")
        return
    out_path.parent.mkdir(parents=True, exist_ok=True)
    if url is None:
        print(f"Creating empty file: {out_path.name}")
        out_path.touch()
        return
    subprocess.run(build_aria2_cmd(url, out_path), check=True)

saved = load_state()

common_downloads = [
    ("Wan VAE", "https://huggingface.co/Comfy-Org/Wan_2.2_ComfyUI_Repackaged/resolve/main/split_files/vae/wan_2.1_vae.safetensors?download=true", paths["vae"] / "wan_2.1_vae.safetensors"),
    ("UMT5 XXL", "https://huggingface.co/Comfy-Org/Wan_2.1_ComfyUI_repackaged/resolve/main/split_files/text_encoders/umt5_xxl_fp8_e4m3fn_scaled.safetensors?download=true", paths["text_encoders"] / "umt5_xxl_fp8_e4m3fn_scaled.safetensors"),
    ("Lightx2v", "https://huggingface.co/Kijai/WanVideo_comfy/resolve/709844db75d2e15582cf204e9a0b5e12b23a35dd/Lightx2v/lightx2v_I2V_14B_480p_cfg_step_distill_rank128_bf16.safetensors?download=true", paths["loras"] / "lightx2v_I2V_14B_480p_cfg_step_distill_rank128_bf16.safetensors"),
    ("Lightx2v-1022-h", "https://huggingface.co/lightx2v/Wan2.2-Distill-Loras/resolve/main/wan2.2_i2v_A14b_high_noise_lora_rank64_lightx2v_4step_1022.safetensors", paths["loras"] / "wan2.2_i2v_A14b_high_noise_lora_rank64_lightx2v_4step_1022.safetensors"),
    ("Lightx2v-1022-l", "https://huggingface.co/lightx2v/Wan2.2-Distill-Loras/resolve/main/wan2.2_i2v_A14b_low_noise_lora_rank64_lightx2v_4step_1022.safetensors", paths["loras"] / "wan2.2_i2v_A14b_low_noise_lora_rank64_lightx2v_4step_1022.safetensors"),
    ("Clip Vision", "https://huggingface.co/OreX/Models/resolve/main/WAN/clip_vision_h.safetensors?download=true", paths["clip_vision"] / "clip_vision_h.safetensors"),
    ("Upscale 4x-AnimeSharp", "https://huggingface.co/utnah/esrgan/resolve/main/4x-AnimeSharp.pth?download=true", paths["upscale_models"] / "4x-AnimeSharp.pth"),
]

gguf_downloads = [
    ("Wan2.2-I2V-A14B-HighNoise-Q2_K", None, paths["diffusion"] / "Wan2.2-I2V-A14B-HighNoise-Q2_K.gguf"),
    ("Wan2.2-I2V-A14B-LowNoise-Q2_K", None, paths["diffusion"] / "Wan2.2-I2V-A14B-LowNoise-Q2_K.gguf"),
]

checkpoint_groups = [
    {"name": "DaSiWa v11", "high": ("https://civitai.red/api/download/models/2953474?fileId=2837908", paths["diffusion"] / "DaSiWa_v11_High.safetensors"), "low": ("https://civitai.red/api/download/models/2953485?fileId=2837910", paths["diffusion"] / "DaSiWa_v11_Low.safetensors")},
    {"name": "SmoothMix Wan 2.2 14B I2V 2.0", "high": ("https://civitai.red/api/download/models/2513182?fileId=2460270", paths["diffusion"] / "SmoothMix_Wan2.2_14B_I2V_2.0_High.safetensors"), "low": ("https://civitai.red/api/download/models/2513186?fileId=2460676", paths["diffusion"] / "SmoothMix_Wan2.2_14B_I2V_2.0_Low.safetensors")},
]

lora_groups = [
    {"name": "rub-pussy", "high": ("https://civitai.red/api/download/models/2563945?fileId=2452277", paths["loras"] / "rub-pussy-high.safetensors"), "low": ("https://civitai.red/api/download/models/2564047?fileId=2452323", paths["loras"] / "rub-pussy-low.safetensors")},
    {"name": "SmoothMix Animations", "high": ("https://civitai.red/api/download/models/2309690?fileId=2499636", paths["loras"] / "smoothmix-animation-high.safetensors"), "low": ("https://civitai.red/api/download/models/2309689?fileId=2210922", paths["loras"] / "smoothmix-animation-low.safetensors")},
    {"name": "SmoothMix Animations XXX", "high": ("https://civitai.red/api/download/models/2376136?fileId=2266910", paths["loras"] / "smoothmix-xxx-animation-high.safetensors"), "low": ("https://civitai.red/api/download/models/2376143?fileId=2266915", paths["loras"] / "smoothmix-xxx-animation-low.safetensors")},
    {"name": "SmoothMix Animations Style", "high": ("https://civitai.red/api/download/models/2318650?fileId=2209031", paths["loras"] / "smoothmix-style-animation-high.safetensors"), "low": ("https://civitai.red/api/download/models/2318707?fileId=2209084", paths["loras"] / "smoothmix-style-animation-low.safetensors")},
    {"name": "Breast Play", "high": ("https://civitai.red/api/download/models/2448064?fileId=2339220", paths["loras"] / "breast-play-high.safetensors"), "low": ("https://civitai.red/api/download/models/2448070?fileId=2339226", paths["loras"] / "breast-play-low.safetensors")},
    {"name": "Bouncy Walk", "high": ("https://civitai.red/api/download/models/2261902?fileId=2154049", paths["loras"] / "bouncy-walk-high.safetensors"), "low": ("https://civitai.red/api/download/models/2261896?fileId=2154043", paths["loras"] / "bouncy-walk-low.safetensors")},
    {"name": "Motion Enhancer - Penis", "high": ("https://civitai.red/api/download/models/2308249?fileId=2199514", paths["loras"] / "penis-high.safetensors"), "low": ("https://civitai.red/api/download/models/2308253?fileId=2199506", paths["loras"] / "penis-low.safetensors")},
    {"name": "Mating Press", "high": ("https://civitai.red/api/download/models/2663475?fileId=2550246", paths["loras"] / "mating-press-high.safetensors"), "low": ("https://civitai.red/api/download/models/2663487?fileId=2550257", paths["loras"] / "mating-press-low.safetensors")},
    {"name": "(wan 2.2 experimental) WAN General NSFW model", "high": ("https://civitai.red/api/download/models/2073605?fileId=1969798", paths["loras"] / "ex-general-nsfw-high.safetensors"), "low": ("https://civitai.red/api/download/models/2083303?fileId=1979213", paths["loras"] / "ex-general-nsfw-low.safetensors")},
    {"name": "WAN 2.2 I2V - Handjob (One/Two handed)", "high": ("https://civitai.red/api/download/models/2352366?fileId=2242793", paths["loras"] / "handjob-1-2-hands-high.safetensors"), "low": ("https://civitai.red/api/download/models/2352388?fileId=2242807", paths["loras"] / "handjob-1-2-hands-low.safetensors")},
    {"name": "Wan2.2-i2v-Masturbation/handjob", "high": ("https://civitai.red/api/download/models/2620366?fileId=2507859", paths["loras"] / "masturbation-handjob-high.safetensors"), "low": ("https://civitai.red/api/download/models/2622170?fileId=2509742", paths["loras"] / "masturbation-handjob-low.safetensors")},
    {"name": "SVI v2 PRO", "high": ("https://huggingface.co/Kijai/WanVideo_comfy/resolve/main/LoRAs/Stable-Video-Infinity/v2.0/SVI_v2_PRO_Wan2.2-I2V-A14B_HIGH_lora_rank_128_fp16.safetensors?download=true", paths["loras"] / "SVI_v2_PRO_Wan2.2-I2V-A14B_HIGH_lora_rank_128_fp16.safetensors"), "low": ("https://huggingface.co/Kijai/WanVideo_comfy/resolve/main/LoRAs/Stable-Video-Infinity/v2.0/SVI_v2_PRO_Wan2.2-I2V-A14B_LOW_lora_rank_128_fp16.safetensors?download=true", paths["loras"] / "SVI_v2_PRO_Wan2.2-I2V-A14B_LOW_lora_rank_128_fp16.safetensors")},
]

common_checks = {name: widgets.Checkbox(value=saved.get("common", {}).get(name, True), description=name, indent=False) for name, _, _ in common_downloads}
checkpoint_radio = widgets.RadioButtons(options=[g["name"] for g in checkpoint_groups], value=saved.get("checkpoint", checkpoint_groups[0]["name"]), description="Checkpoint:")
lora_checks = {g["name"]: widgets.Checkbox(value=saved.get("loras", {}).get(g["name"], False), description=g["name"], indent=False) for g in lora_groups}

btn_download = widgets.Button(description="Download selected", button_style="success")
btn_save = widgets.Button(description="Save selection", button_style="info")
btn_all_loras = widgets.Button(description="Select all LoRA")
btn_none_loras = widgets.Button(description="Clear all LoRA")
out = widgets.Output()


def on_save(_):
    save_state(common_checks, checkpoint_radio, lora_checks)
    with out:
        clear_output()
        print(f"Saved to {state_file}")


def on_download(_):
    save_state(common_checks, checkpoint_radio, lora_checks)
    with out:
        clear_output()
        for name, url, out_path in common_downloads:
            if common_checks[name].value:
                print(f"\nDownloading {name} -> {out_path.name}")
                download_one(url, out_path)
        for name, url, out_path in gguf_downloads:
            print(f"\nPreparing GGUF file {name} -> {out_path.name}")
            download_one(url, out_path)
        for g in checkpoint_groups:
            if g["name"] == checkpoint_radio.value:
                print(f"\nDownloading {g['name']} [high] -> {g['high'][1].name}")
                download_one(g["high"][0], g["high"][1])
                print(f"\nDownloading {g['name']} [low] -> {g['low'][1].name}")
                download_one(g["low"][0], g["low"][1])
                break
        for g in lora_groups:
            if lora_checks[g["name"]].value:
                print(f"\nDownloading {g['name']} [high] -> {g['high'][1].name}")
                download_one(g["high"][0], g["high"][1])
                print(f"\nDownloading {g['name']} [low] -> {g['low'][1].name}")
                download_one(g["low"][0], g["low"][1])


def on_select_all_loras(_):
    for cb in lora_checks.values():
        cb.value = True


def on_clear_all_loras(_):
    for cb in lora_checks.values():
        cb.value = False

btn_save.on_click(on_save)
btn_download.on_click(on_download)
btn_all_loras.on_click(on_select_all_loras)
btn_none_loras.on_click(on_clear_all_loras)

display(widgets.VBox([
    widgets.Label("Common downloads:"),
    widgets.VBox(list(common_checks.values())),
    widgets.Label("GGUF placeholders (empty files):"),
    widgets.VBox([widgets.Label(f"- {name}") for name, _, _ in gguf_downloads]),
    widgets.Label("Checkpoint (one model, downloads high + low):"),
    checkpoint_radio,
    widgets.Label("LoRA models (one model, downloads high + low):"),
    widgets.VBox(list(lora_checks.values())),
    widgets.HBox([btn_all_loras, btn_none_loras]),
    widgets.HBox([btn_download, btn_save]),
    out,
]))
```
